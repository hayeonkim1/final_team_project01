```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# 1. 긍정 키워드 및 빈도수 데이터 세팅
positive_words = {
    '좋아요': 47,
    '편리해요': 30,
    '안심': 18,
    '깨끗해요': 16,
    '청결/깨끗': 15,
    '가성비': 12,
    '만족스러워요': 11,
    '친절해요': 7,
    '쾌적': 6,
    '최고예요': 5,
    '신세계': 3,
    '추천': 3,
    '간편해요': 2,
    '안전해요': 2
}

# 2. 한글 폰트 경로 지정 (환경에 맞게 수정 필요)
# Windows: 'C:/Windows/Fonts/malgun.ttf' (맑은고딕) 또는 'C:/Windows/Fonts/NanumGothic.ttf'
# Mac: '/System/Library/Fonts/Supplemental/AppleGothic.ttf'
font_path = 'C:/Windows/Fonts/malgun.ttf'

# 3. 워드클라우드 객체 생성
wordcloud = WordCloud(
    font_path=font_path,
    background_color='white',  # 배경 색상
    width=800,                 # 이미지 가로 크기
    height=600,                # 이미지 세로 크기
    colormap='winter',         # 긍정 느낌의 푸른색조 테마
    random_state=42            # 실행할 때마다 단어 배치가 바뀌지 않도록 고정
).generate_from_frequencies(positive_words)

# 4. 화면 출력 및 시각화 설정
plt.figure(figsize=(10, 8))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # 축 눈금 숨기기

# 5. 이미지 파일로 저장하고 싶다면 아래 주석을 해제하세요
plt.savefig('positive_wordcloud.png', bbox_inches='tight')

plt.show()
```

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# 1. 부정 키워드 및 빈도수 데이터 세팅
negative_words = {
    '고객센터': 11,
    '오류/에러': 9,
    '안 돼요': 7,
    '불편해요': 6,
    '느려요': 4,
    '답답해요': 3,
    '화나요/짜증나요': 3,
    '불친절해요': 2,
    '아쉬워요': 2
}

# 2. 한글 폰트 경로 지정 (환경에 맞게 수정 필요)
font_path = 'C:/Windows/Fonts/malgun.ttf'

# 3. 워드클라우드 객체 생성
wordcloud = WordCloud(
    font_path=font_path,
    background_color='white',  # 배경 색상
    width=800,                 # 이미지 가로 크기
    height=600,                # 이미지 세로 크기
    colormap='autumn',         # 경고/불만 느낌의 붉은·오렌지색조 테마
    random_state=42            # 배치 고정
).generate_from_frequencies(negative_words)

# 4. 화면 출력 및 시각화 설정
plt.figure(figsize=(10, 8))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')

# 5. 이미지 파일로 저장하고 싶다면 아래 주석을 해제하세요
plt.savefig('negative_wordcloud.png', bbox_inches='tight')

plt.show()
```
