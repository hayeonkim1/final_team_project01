## 0. 분석 목적 및 이유

### 왜 이 분석을 하는가?

짐토리의 **메인 타겟은 20대 대학생 1인가구**입니다. 그러나 성장을 위해서는 **대학생이 아닌 세그먼트** 중 어디로 확장할지 STP 관점에서 근거가 필요합니다.

### 핵심 질문

1. 어떤 **feature 조합**이 서비스 선호(Q6)가 가장 높은가?
2. **비대학생** 중 확장 1순위 직업·가구·행동 패턴은 무엇인가?
3. 이진(선호/비선호)이 아닌 **Q6 연속 척도**로 예측·실측을 같은 눈금에서 비교할 수 있는가?

### 방법론 선택 이유

| 선택 | 이유 |
|------|------|
| **실측 데이터 (10대 제외)** | 10대 1명은 표본이 극소·외삽 위험 → 분석 일관성 |
| **Q6 가중 확률** `(Q6÷5)×100` | 4점·5점·3점의 차이를 반영, 예측·실측 동일 척도 |
| **RF 회귀 (Q6 직접 예측)** | 분류 `predict_proba`는 0/1에 쏠림 → Q6 회귀로 연속 의향 예측 |
| **Feature 5개 조합 평균** | 개인 1명이 아닌 **세그먼트 프로필** 단위로 확장 가능성 비교 |
| **전수탐색 (960만 조합)** | 이론적 최적 조합의 상한선 파악 (가상 프로필) |

$$\text{선호확률(\%)} = \frac{\text{Q6 점수}}{5} \times 100$$

| Q6 | 가중 선호확률 | 해석 |
|----|-------------|------|
| 5 | 100% | 매우 그렇다 |
| 4 | 80% | 그렇다 (기존 이진 선호 기준) |
| 3 | 60% | 보통 |
| 2 | 40% | 아니다 |
| 1 | 20% | 전혀 아니다 |

## 3. RF 모델 Feature 조합 전수탐색

14개 입력 feature의 **모든 값 조합**(약 **960만**개)에 대해 RF 회귀 예측 Q6 → 선호확률을 계산하고, **상위 Top 5** 조합을 추출합니다.

> ⚠️ 가상 프로필 전수탐색 → 설문에 없는 조합(예: 10대+자영업자)도 포함될 수 있음.  
> `RUN_GRID_SEARCH=False`이면 저장된 CSV를 로드합니다 (약 4분 소요).

```python
RUN_GRID_SEARCH = True # True로 변경 시 전수탐색 실행 (~4분)
TOP_K = 5
BATCH_SIZE = 100_000

cat_cols, num_cols = get_feature_columns(use_behavioral=True)
keys = cat_cols + num_cols

pipe_grid = make_pipeline(
    cat_cols, num_cols,
    RandomForestRegressor(n_estimators=200, random_state=42),
)
pipe_grid.fit(weighted_all[keys], weighted_all["q6_service_intent"])

cat_domains = {c: sorted(features.loc[mask_all, c].dropna().unique().tolist()) for c in cat_cols}
num_domains = {
    "move_frequency": [0, 1, 2, 3, 4],
    "storage_need_score": [1, 2, 3, 4, 5],
    "pain_score": [1, 2, 3, 4, 5],
    "trust_score": [1, 2, 3, 4, 5],
    "has_rental_experience": [0, 1],
    "consignment_positive": [0, 1],
    "is_single_household": [0, 1],
    "pain_transport": [0, 1],
    "pain_space": [0, 1],
    "pain_cost": [0, 1],
}
domains = [cat_domains[c] for c in cat_cols] + [num_domains[c] for c in num_cols]
grid_size = int(np.prod([len(d) for d in domains]))
print(f"Grid size: {grid_size:,}")

grid_csv = OUTPUT_DIR / "top5_feature_combinations.csv"

if RUN_GRID_SEARCH:
    heap = []
    counter = 0
    batch = []
    for combo in product(*domains):
        batch.append(dict(zip(keys, combo)))
        if len(batch) >= BATCH_SIZE:
            Xb = pd.DataFrame(batch)
            q6_pred = np.clip(pipe_grid.predict(Xb), 1, 5)
            for row, q6 in zip(batch, q6_pred):
                p = float(q6_to_prob_pct(q6))
                item = (p, counter, {**row, "predicted_q6_score": round(q6, 2), "predicted_prob_pct": round(p, 2)})
                counter += 1
                if len(heap) < TOP_K:
                    heapq.heappush(heap, item)
                elif p > heap[0][0]:
                    heapq.heapreplace(heap, item)
            batch = []
    if batch:
        Xb = pd.DataFrame(batch)
        q6_pred = np.clip(pipe_grid.predict(Xb), 1, 5)
        for row, q6 in zip(batch, q6_pred):
            p = float(q6_to_prob_pct(q6))
            item = (p, counter, {**row, "predicted_q6_score": round(q6, 2), "predicted_prob_pct": round(p, 2)})
            counter += 1
            if len(heap) < TOP_K:
                heapq.heappush(heap, item)
            elif p > heap[0][0]:
                heapq.heapreplace(heap, item)
    grid_top5 = pd.DataFrame([x[2] for x in sorted(heap, key=lambda t: t[0], reverse=True)])
    grid_top5.insert(0, "rank", range(1, len(grid_top5) + 1))
    grid_top5.to_csv(grid_csv, index=False, encoding="utf-8-sig")
    print(f"Saved: {grid_csv}")
else:
    if grid_csv.exists():
        grid_top5 = pd.read_csv(grid_csv, encoding="utf-8-sig")
        print(f"Loaded cached: {grid_csv}")
    else:
        print("캐시 없음 → RUN_GRID_SEARCH=True 로 실행하세요.")
        grid_top5 = pd.DataFrame()

if not grid_top5.empty:
    display(grid_top5[["rank", "occupation", "age_group", "housing_type",
                       "storage_need_score", "pain_score", "trust_score", "predicted_prob_pct"]])
```

## 5. 비대학생 — 5-Feature 조합별 평균 상·하위 Top5

### 조합 정의

동일한 5개 feature 값을 가진 응답자끼리 그룹화하여 **평균** 선호확률 산출:

`occupation` · `housing_type` · `is_single_household` · `pain_score` · `trust_score`

- 최소 표본: **n ≥ 2**
- 모델: 전체(10대 제외) 학습 → **비대학생만** 스코어링

```python
def aggregate_combo_means(result, combo_features, min_n=2):
    g = (
        result.groupby(combo_features, dropna=False)
        .agg(
            n=("predicted_prob_pct", "size"),
            avg_predicted_prob_pct=("predicted_prob_pct", "mean"),
            avg_actual_prob_pct=("actual_prob_pct", "mean"),
        )
        .reset_index()
    )
    g = g[g["n"] >= min_n].copy()
    g["avg_predicted_prob_pct"] = g["avg_predicted_prob_pct"].round(2)
    g["avg_actual_prob_pct"] = g["avg_actual_prob_pct"].round(2)
    g["combo_label"] = g.apply(
        lambda r: " | ".join(f"{c}={r[c]}" for c in combo_features), axis=1
    )
    return g

weighted_non_student, _ = build_weighted_result(
    df, features, targets,
    exclude_age=EXCLUDE_AGE,
    exclude_occupations=["대학생"],
    fit_scope="all",
)

combo_all = aggregate_combo_means(weighted_non_student, COMBO_FEATURES)
combo_top5 = combo_all.sort_values("avg_predicted_prob_pct", ascending=False).head(5).copy()
combo_bottom5 = combo_all.sort_values("avg_predicted_prob_pct", ascending=True).head(5).copy()
combo_top5.insert(0, "rank", range(1, 6))
combo_bottom5.insert(0, "rank", range(1, 6))

print(f"비대학생 n={len(weighted_non_student)}, 유효 조합(n>=2): {len(combo_all)}개\n")
print("=== 비대학생 상위 Feature 조합 Top 5 (평균) ===")
display(combo_top5[["rank", "occupation", "housing_type", "is_single_household",
                    "pain_score", "trust_score", "n", "avg_predicted_prob_pct", "avg_actual_prob_pct"]])
print("\n=== 비대학생 하위 Feature 조합 Top 5 (평균) ===")
display(combo_bottom5[["rank", "occupation", "housing_type", "is_single_household",
                       "pain_score", "trust_score", "n", "avg_predicted_prob_pct", "avg_actual_prob_pct"]])

combo_top5.to_csv(OUTPUT_DIR / "top5_feature_combo_means_non_student.csv", index=False, encoding="utf-8-sig")
combo_bottom5.to_csv(OUTPUT_DIR / "bottom5_feature_combo_means_non_student.csv", index=False, encoding="utf-8-sig")
```

```python
occ_summary = (
    weighted_non_student.groupby("occupation")
    .agg(
        n=("predicted_prob_pct", "size"),
        avg_predicted=("predicted_prob_pct", "mean"),
        avg_actual=("actual_prob_pct", "mean"),
    )
    .reset_index()
    .sort_values("avg_predicted", ascending=False)
)
occ_summary["avg_predicted"] = occ_summary["avg_predicted"].round(2)
occ_summary["avg_actual"] = occ_summary["avg_actual"].round(2)

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("직업별 평균 선호확률 (비대학생)", "Feature 조합 Top/Bottom 5"),
    specs=[[{"type": "bar"}, {"type": "bar"}]],
    column_widths=[0.45, 0.55],
)
fig.add_trace(
    go.Bar(x=occ_summary["avg_predicted"], y=occ_summary["occupation"], orientation="h",
           name="예측", marker_color="#4C78A8"),
    row=1, col=1,
)
fig.add_trace(
    go.Bar(x=occ_summary["avg_actual"], y=occ_summary["occupation"], orientation="h",
           name="실측", marker_color="#72B7B2"),
    row=1, col=1,
)

combo_plot = pd.concat([
    combo_top5.assign(group="상위 조합"),
    combo_bottom5.assign(group="하위 조합"),
])
combo_plot["short_label"] = combo_plot["occupation"] + " / " + combo_plot["housing_type"].str[:4]
fig.add_trace(
    go.Bar(
        x=combo_plot["avg_predicted_prob_pct"], y=combo_plot["short_label"],
        orientation="h", marker_color=combo_plot["group"].map({"상위 조합": "#4C78A8", "하위 조합": "#E45756"}),
        showlegend=False,
    ),
    row=1, col=2,
)
fig.update_layout(title="비대학생 세그먼트 확장 — 직업·조합별 Q6 가중 선호확률", template=PLOTLY_TEMPLATE, height=520, barmode="group")
fig.show()

display(occ_summary)
```

### 📌 5단계 소결론 (비대학생 Feature 조합)

#### 상위 조합 공통점
- **1인 가구** (`is_single_household=1`)
- **페인5** (또는 4)
- 직업: **직장인·취준생·자영업자** — 1인가구 + 고페인이면 고확률

#### 하위 조합 공통점
- **4인+·2인 가구**, **페인·신뢰 낮음**
- 최저: 취준생 + 4인+가구 + 페인3 + 신뢰2 → **평균 63.4%**

|------|------|---|----------|----------|

## 6. 최종 결론

### 분석 요약

1. **실측 105명(10대 제외)** + **Q6 가중 선호확률**로 예측·실측을 동일 척도(0~100%)에서 비교.
2. **RF 회귀**가 Q6를 직접 예측 → feature 조합 **전수탐색(960만)** 및 **그룹 평균** 분석에 활용.
3. 전수탐색 상한: *1인가구 + 최고 니즈·페인·신뢰* (가상 프로필 100%).
4. **비대학생** 확장은 개인이 아닌 **5-feature 조합 평균**으로 판단하는 것이 STP에 적합.
