```python
from google_play_scraper import Sort, reviews
import pandas as pd
from datetime import datetime

app_id = 'com.secondsyndrome.dalock'  # 예: 'com.nhn.android.naver'

# 설정할 날짜 기준 (타임존 제외한 날짜 비교를 위해 tz_localize 처리 예정)
start_date = pd.to_datetime('2025-06-01')
end_date = pd.to_datetime('2026-06-18')

print("구글 플레이스토어 리뷰 수집 시작...")

all_reviews = []
continuation_token = None

while True:
    # 한번에 200개씩 최신순으로 가져옴
    result, continuation_token = reviews(
        app_id,
        lang='ko',
        country='kr',
        sort=Sort.NEWEST,
        count=200,
        continuation_token=continuation_token
    )
    
    if not result:
        break
        
    all_reviews.extend(result)
    
    # 마지막으로 가져온 리뷰의 날짜 확인
    last_review_date = pd.to_datetime(result[-1]['at']).numerator # 타임존 제거 비교용
    last_review_date = pd.to_datetime(result[-1]['at']).replace(tzinfo=None)
    
    # 만약 마지막 리뷰 날짜가 우리가 원하는 시작일(2025-06-01)보다 과거라면 루프 종료
    if last_review_date < start_date:
        print("설정한 수집 시작일 이전 데이터에 도달하여 수집을 중단합니다.")
        break

# 데이터프레임 변환
df_google = pd.DataFrame(all_reviews)

# 날짜 컬럼을 비교 가능한 datetime 형태로 변환 (타임존 제거)
df_google['at'] = pd.to_datetime(df_google['at']).dt.tz_localize(None)

# 지정한 기간 (2025-06-01 ~ 2026-06-18) 데이터만 필터링
df_google_filtered = df_google[(df_google['at'] >= start_date) & (df_google['at'] <= end_date)].copy()

# 필요한 컬럼만 선택 및 이름 변경
df_google_filtered = df_google_filtered[['userName', 'score', 'content', 'at']]
df_google_filtered.columns = ['작성자', '별점', '리뷰내용', '작성일']

# CSV 저장 (utf-8-sig 반영)
df_google_filtered.to_csv('google_play_reviews_filtered.csv', index=False, encoding='utf-8-sig')
print(f"구글 리뷰 저장 완료! 조건에 맞는 리뷰 총 {len(df_google_filtered)}개 저장됨.")
```

```python
import requests
import pandas as pd
from datetime import datetime

# 1. 앱 정보 설정 (예시: 당근 ID - 1018769995)
# 숫자로 된 앱 ID만 정확하게 입력해주세요!
app_id = 1633014747 

# 설정할 날짜 기준
start_date = pd.to_datetime('2025-06-01')
end_date = pd.to_datetime('2026-06-18')

print("애플 앱스토어 API 직접 수집 시작...")

all_reviews = []

# 애플 공식 RSS 피드는 페이지당 50개씩, 최대 10페이지(500개)까지 제공합니다.
for page in range(1, 11):
    url = f"https://itunes.apple.com/kr/rss/customerreviews/page={page}/id={app_id}/sortby=mostrecent/json"
    
    response = requests.get(url)
    if response.status_code != 200:
        print(f"{page}페이지를 가져오는 데 실패했습니다.")
        break
        
    data = response.json()
    
    # 리뷰 데이터가 없는 경우 종료
    feed = data.get('feed', {})
    entries = feed.get('entry', [])
    
    if not entries:
        break
        
    # 첫 번째 엔트리가 앱 소개 정보인 경우가 있어 예외 처리
    if isinstance(entries, dict):
        entries = [entries]
        
    for entry in entries:
        # 앱 자체 정보 피드는 건너뛰기
        if 'im:name' in entry:
            continue
            
        try:
            # 날짜 파싱 및 타임존 제거
            raw_date = entry['updated']['label']
            review_date = pd.to_datetime(raw_date).tz_localize(None)
            
            # 기간 필터링 적용
            if start_date <= review_date <= end_date:
                all_reviews.append({
                    '작성자': entry['author']['name']['label'],
                    '별점': int(entry['im:rating']['label']),
                    '리뷰내용': entry['content']['label'],
                    '작성일': review_date,
                    '리뷰제목': entry['title']['label']
                })
        except KeyError:
            continue

# 2. 데이터프레임 변환 및 저장
if all_reviews:
    df_apple = pd.DataFrame(all_reviews)
    
    # 중복 제거 및 정렬
    df_apple = df_apple.drop_duplicates(subset=['작성자', '리뷰내용'])
    df_apple = df_apple.sort_values(by='작성일', ascending=False)
    
    # CSV 저장 (utf-8-sig 반영)
    df_apple.to_csv('app_store_reviews_filtered.csv', index=False, encoding='utf-8-sig')
    print(f"애플 리뷰 저장 완료! 조건에 맞는 리뷰 총 {len(df_apple)}개 저장됨.")
else:
    print("해당 기간(2025-06-01 ~ 2026-06-18) 내에 작성된 리뷰가 없거나 앱 ID를 확인해주세요.")
```

```python
from urllib.request import Request, urlopen
import xml.etree.ElementTree as ET
import pandas as pd

# 1. 실제 다락 앱의 앱스토어 ID 숫자 입력 (예: 123456789)
app_id = 1633014747 # 다락 앱의 실제 ID 숫자로 변경해 주세요!

# 설정할 날짜 기준
start_date = pd.to_datetime('2025-06-01')
end_date = pd.to_datetime('2026-06-18')

print("애플 앱스토어 XML 엔진으로 수집 시작...")

all_reviews = []
# XML 파싱을 위한 네임스페이스 정의
ns = {
    'im': 'http://itunes.apple.com/rss',
    'feed': 'http://www.w3.org/2005/Atom'
}

for page in range(1, 11):
    # 리뷰 수가 적을 때는 xml 형식이 훨씬 정확하고 안정적입니다.
    url = f"https://itunes.apple.com/kr/rss/customerreviews/page={page}/id={app_id}/sortby=mostrecent/xml"
    
    try:
        req = Request(url, headers={'User-Agent': 'Mozilla/5.0'})
        with urlopen(req) as response:
            if response.status != 200:
                break
            xml_data = response.read()
            
        root = ET.fromstring(xml_data)
        entries = root.findall('feed:entry', ns)
        
        if not entries:
            break
            
        for entry in entries:
            # 앱 정보가 들어있는 첫 번째 엔트리는 제외
            if entry.find('im:name', ns) is not None:
                continue
                
            try:
                # 데이터 추출
                raw_date = entry.find('feed:updated', ns).text
                review_date = pd.to_datetime(raw_date).tz_localize(None)
                
                # 기간 필터링 적용 (2025-06-01 ~ 2026-06-18)
                if start_date <= review_date <= end_date:
                    author = entry.find('feed:author/feed:name', ns).text
                    rating = int(entry.find('im:rating', ns).text)
                    title = entry.find('feed:title', ns).text
                    content = entry.find('feed:content[@type="text"]', ns).text
                    
                    all_reviews.append({
                        '작성자': author,
                        '별점': rating,
                        '리뷰내용': content,
                        '작성일': review_date,
                        '리뷰제목': title
                    })
            except AttributeError:
                continue
    except Exception as e:
        print(f"{page}페이지 처리 중 오류 발생: {e}")
        break

# 2. 데이터프레임 변환 및 저장
if all_reviews:
    df_apple = pd.DataFrame(all_reviews)
    
    # 중복 제거 및 정렬
    df_apple = df_apple.drop_duplicates(subset=['작성자', '리뷰내용'])
    df_apple = df_apple.sort_values(by='작성일', ascending=False)
    
    # CSV 저장 (utf-8-sig 반영)
    df_apple.to_csv('app_store_reviews_filtered.csv', index=False, encoding='utf-8-sig')
    print(f"애플 리뷰 저장 완료! 조건에 맞는 리뷰 총 {len(df_apple)}개 저장됨.")
else:
    print("지정한 기간(2025-06-01 ~ 2026-06-18) 내에 작성된 리뷰가 없거나 앱 ID를 다시 확인해 주세요.")
```

```python
from urllib.request import Request, urlopen
import xml.etree.ElementTree as ET
import pandas as pd

# '다락' 앱의 실제 ID 숫자 (코드가 맞다면 그대로 두셔도 됩니다)
app_id = 1633014747  

print("애플 앱스토어 (전체 기간) 수집 시작...")

all_reviews = []
ns = {
    'im': 'http://itunes.apple.com/rss',
    'feed': 'http://www.w3.org/2005/Atom'
}

for page in range(1, 11):
    url = f"https://itunes.apple.com/kr/rss/customerreviews/page={page}/id={app_id}/sortby=mostrecent/xml"
    
    try:
        req = Request(url, headers={'User-Agent': 'Mozilla/5.0'})
        with urlopen(req) as response:
            if response.status != 200: break
            xml_data = response.read()
            
        root = ET.fromstring(xml_data)
        entries = root.findall('feed:entry', ns)
        
        if not entries: break
            
        for entry in entries:
            if entry.find('im:name', ns) is not None: continue
                
            try:
                raw_date = entry.find('feed:updated', ns).text
                review_date = pd.to_datetime(raw_date).tz_localize(None)
                
                # 💡 기간 필터 전면 제거! 들어오는 대로 다 넣습니다.
                author = entry.find('feed:author/feed:name', ns).text
                rating = int(entry.find('im:rating', ns).text)
                title = entry.find('feed:title', ns).text
                content = entry.find('feed:content[@type="text"]', ns).text
                
                all_reviews.append({
                    '작성자': author,
                    '별점': rating,
                    '리뷰내용': content,
                    '작성일': review_date,
                    '리뷰제목': title
                })
            except AttributeError:
                continue
    except Exception as e:
        print(f"오류 발생: {e}")
        break

# 2. 데이터프레임 변환 및 저장
if all_reviews:
    df_apple = pd.DataFrame(all_reviews)
    df_apple = df_apple.drop_duplicates(subset=['작성자', '리뷰내용']).sort_values(by='작성일', ascending=False)
    
    # 파일명을 다르게 저장합니다.
    df_apple.to_csv('app_store_reviews_ALL.csv', index=False, encoding='utf-8-sig')
    print(f"🎉 수집 완료! 존재하느 모든 리뷰 총 {len(df_apple)}개가 'app_store_reviews_ALL.csv'에 저장되었습니다.")
    print(f"📌 수집된 리뷰 중 가장 최신 리뷰의 날짜: {df_apple['작성일'].max()}")
else:
    print("❌ 이 앱은 앱스토어에 등록된 리뷰가 전 기간을 통틀어 '0개'입니다.")
```

```python
import urllib.request
from bs4 import BeautifulSoup
import pandas as pd

# 보내주신 미니창고 다락의 실제 앱스토어 웹 주소
url = "https://apps.apple.com/kr/app/%EB%AF%B8%EB%8B%88%EC%B0%BD%EA%B3%A0-%EB%8B%A4%EB%9D%BD/id1633014747"

# 설정할 날짜 기준 (2025-06-01 ~ 2026-06-18)
start_date = pd.to_datetime('2025-06-01')
end_date = pd.to_datetime('2026-06-18')

print("애플 웹페이지 직접 크롤링 시작...")

try:
    # 브라우저인 것처럼 속여서 페이지 요청
    req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'})
    html = urllib.request.urlopen(req).read()
    soup = BeautifulSoup(html, 'html.parser')
    
    # 웹 페이지 내의 리뷰 컴포넌트 추출
    review_elements = soup.select('.we-customer-review')
    
    all_reviews = []
    
    for element in review_elements:
        try:
            # 1. 날짜 추출 및 변환
            date_str = element.select_one('.we-customer-review__date').text.strip()
            # 앱스토어 날짜 형식(예: 2023. 11. 21.)을 datetime으로 변환
            review_date = pd.to_datetime(date_str, format='%Y. %m. %d.').tz_localize(None)
            
            # 2. 기간 필터링 적용
            if start_date <= review_date <= end_date:
                author = element.select_one('.we-customer-review__user').text.strip()
                title = element.select_one('.we-customer-review__title').text.strip()
                content = element.select_one('.we-customer-review__body').text.strip()
                
                # 별점 추출 (aria-label="5점 만점에 5점" 등에서 숫자만 추출)
                rating_attr = element.select_one('.we-star-rating-badge')['aria-label']
                rating = int(rating_attr.split('점')[0].strip()) # 앞의 숫자만 가져옴
                
                all_reviews.append({
                    '작성자': author,
                    '별점': rating,
                    '리뷰내용': content,
                    '작성일': review_date,
                    '리뷰제목': title
                })
        except Exception:
            continue

    # 3. 결과 저장
    if all_reviews:
        df_apple = pd.DataFrame(all_reviews)
        df_apple = df_apple.sort_values(by='작성일', ascending=False)
        
        # CSV 저장
        df_apple.to_csv('app_store_reviews_filtered.csv', index=False, encoding='utf-8-sig')
        print(f"🎉 성공! 지정 기간 내 리뷰 총 {len(df_apple)}개가 'app_store_reviews_filtered.csv'에 저장되었습니다.")
    else:
        print("⚠️ 웹페이지는 정상 조회되었으나, 지정하신 기간(2025-06-01 ~ 2026-06-18) 내에 작성된 리뷰가 페이지에 없습니다.")
        
except Exception as e:
    print(f"크롤링 중 에러 발생: {e}")
```
